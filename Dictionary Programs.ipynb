```python
res={101:"Amit",102:"Rahul",103:"Priya"}
key=int(input("Enter key to search: "))
if key in res:
    print("The key exists in dictionary")
else:
    print("The key does not exist in dictionary")
```

```python
res={101:"Amit",102:"Rahul",103:"Priya"}
print("Keys: \n")
for key in res:
    print(key)
print("\n Values")
for value in res.values():
    print(value)
print("\nKey and values:", res)
```

```python
num={"a":10,"b":20,"c":30,"d":40,}
s=0
for value in num.values():
    s = s+value
print("Sum is: ", s)
```

```python
stdinfo={101:"Amit",102:"Rahul",103:"Priya",104:"Jane",105:"Jason"}
del stdinfo[102]
stdinfo[102]= "Harry"
stdinfo=dict(sorted(stdinfo.items()))
print(stdinfo)
```

```python
l1=["name", "panel", "rollno"]
l2=["ABC","B",34]
dict1=dict(zip(l1,l2))
print("Dictionary: ", dict1)
```

```python
stdinfo={101:"Amit",102:"Rahul",103:"Priya",104:"Jane",105:"Jason"}
l1=[]
l2=[]
for key in stdinfo:
    l1.append(key)
for value in stdinfo.values():
    l2.append(value)
print("Keys in list 1: ", l1)
print("Values in list 2: ", l2)
```

```python
num={"a":10,"b":20,"c":30,"d":40,}
s=0
i=0
for value in num.values():
    s=s+value
    i=i+1
avg = s/i
print("Avg = ", avg)
```
